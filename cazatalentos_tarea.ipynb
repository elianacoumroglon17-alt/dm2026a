Loa cazatalentos 3, 6 y 7 indican resultados que no están influenciados por el overfitting ya que no se comparan con un grupo.

Para C1, C2 y C4, los casos que sí están influenciados por el grupo, me pregunto:

Si el vardadero índice de la jugadora fuese 0.65, 0.66, 0.67, etc. hasta 0.8, cuántos tiros libres acertaría esa jugadora en un grupo de x participantes (100 para el caso de C1, 200 para el caso de C2 y 2 para el caso de C4)

```r
set.seed(102191)

# Funcion base
esperanza_del_maximo <- function(p, n_jugadoras, tiros, n_sim = 5000) {
  maximos <- replicate(n_sim, max(rbinom(n_jugadoras, size = tiros, prob = p)))
  mean(maximos)
}

# C1: ganadora de pool de 100
cat("=== C1 (pool 100) ===\n")
for (p in seq(0.65, 0.80, by = 0.01)) {
  cat("p =", p, "-> maximo esperado =",
      esperanza_del_maximo(p, n_jugadoras=100, tiros=100), "\n")
}

# C2: ganadora de pool de 200
cat("\n=== C2 (pool 200) ===\n")
for (p in seq(0.65, 0.80, by = 0.01)) {
  cat("p =", p, "-> maximo esperado =",
      esperanza_del_maximo(p, n_jugadoras=200, tiros=100), "\n")
}

# C4: ganadora de pool de 2
cat("\n=== C4 (pool 2) ===\n")
for (p in seq(0.65, 0.80, by = 0.01)) {
  cat("p =", p, "-> maximo esperado =",
      esperanza_del_maximo(p, n_jugadoras=2, tiros=100), "\n")
}
```
