Para C1, C2 y C4, los casos que sí están influenciados por el grupo, me pregunto:

A mayor tamaño de la muestra más overfitting, por lo tanto esperaría que el orden en este caso sea C2 < C1 < C4.

```r
set.seed(102191)
n_experimentos <- 100000
tiros <- 100

indice_c1 <- c()
indice_c2 <- c()
indice_c4 <- c()

# Mismo universo de jugadoras
tamano_grupo <- function(n) {
  return(runif(n, 0.45, 0.75))
}

for (i in 1:n_experimentos) {

  # CAZATALENTOS 1: Grupo de 100, pero la ganadora metió EXACTAMENTE 80
  p_grupo1 <- tamano_grupo(100)
  tiroslibres1    <- rbinom(100, size = tiros, prob = p_grupo1)
  if (max(tiroslibres1) == 80) {
    indice_c1 <- c(indice_c1, p_grupo1[which.max(tiroslibres1)])
  }

  # CAZATALENTOS 2: Grupo de 200, pero la ganadora metió EXACTAMENTE 80
  p_grupo2 <- tamano_grupo(200)
  tiroslibres2    <- rbinom(200, size = tiros, prob = p_grupo2)
  if (max(tiroslibres2) == 80) {
    indice_c2 <- c(indice_c2, p_grupo2[which.max(tiroslibres2)])
  }

  # CAZATALENTOS 4: Grupo de 2, pero la ganadora metió EXACTAMENTE 80
  p_grupo4 <- tamano_grupo(2)
  tiroslibres4    <- rbinom(2, size = tiros, prob = p_grupo4)
  if (max(tiroslibres4) == 80) {
    indice_c4 <- c(indice_c4, p_grupo4[which.max(tiroslibres4)])
  }
}

indice_c1_real <- mean(indice_c1)
indice_c2_real <- mean(indice_c2)
indice_c4_real <- mean(indice_c4)

indice_c1_real
indice_c2_real
indice_c4_real
```

Vemos que los resultados de los cazatalentos 1, 2 y 4 son casí idénticos. Esto se debe a que dentro de estos grupos son todas jugadoras "promedio".
